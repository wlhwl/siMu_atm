```python
def hex_string_to_decimal_array(hex_string):
    # Ensure the length of the hex string is a multiple of 3
    if len(hex_string) % 3 != 0:
        raise ValueError("The length of the hex string should be a multiple of 3.")

    decimal_array = []
    
    for i in range(0, len(hex_string), 3):
        hex_value = hex_string[i:i+3]
        decimal_value = int(hex_value, 16)
        decimal_array.append(decimal_value)
    
    return decimal_array

# Example usage
hex_string = "73973a76d76876b76c76776f74f6e568f5f559052a4b74634173da39736334e3173022f32ee2d52d62d92d02ec2e82f430531832a33f34536a3713983a03ca3d03f640042643d44546a4714964ab4b04d64eb4f551052953454555956357458359e59f5ab5b65c55d15ef5ea5f760161c61862263e63964065b65566e66767d67468b68269869e6946ac6a06b86bd6b26ca6ce6c46d86dc6d36e66ea6ee6e16f76f96fe6f270570870d70f70271671871b71e71172772a72d72073273473873873a73d73e73074074174374474674874a74974c74e74f741750752752752754754755755755755754757758757758759758755756"

decimal_array = hex_string_to_decimal_array(hex_string)
print(decimal_array)
```

```python
import pandas as pd

def hex_to_dec(hex_str):
    return int(hex_str, 16)

def parse_12bit_readings(hex_str):
    readings = []
    for i in range(0, len(hex_str), 3):
        reading = hex_to_dec(hex_str[i:i+3])
        readings.append(reading)
    return readings

def parse_data(data):
    index = 0
    results = []

    while index < len(data):
        if data[index:index+8] == "20040601":
            index += 8
            timestamp = hex_to_dec(data[index:index+16])
            index += 16
            event_id = hex_to_dec(data[index:index+8])
            index += 8

            fragment = {
                "timestamp": timestamp,
                "event_id": event_id
            }

            readings_columns = []
            for i in range(1, 8):
                prefix = f"{i:08x}"
                if data[index:index+8] == prefix:
                    index += 8
                    next_prefix = f"{i+1:08x}"
                    next_index = data.find(next_prefix, index)
                    if next_index == -1:
                        readings_data = data[index:]
                    else:
                        readings_data = data[index:next_index]

                    fragment[f"readings_{i}"] = parse_12bit_readings(readings_data)
                    readings_columns.append(f"readings_{i}")
                    index = next_index

            results.append(fragment)
        else:
            index += 1

    return results, readings_columns

def to_dataframe(parsed_data, readings_columns):
    data_dict = {
        "timestamp": [],
        "event_id": []
    }
    
    for col in readings_columns:
        data_dict[col] = []

    for fragment in parsed_data:
        data_dict["timestamp"].append(fragment["timestamp"])
        data_dict["event_id"].append(fragment["event_id"])
        
        for col in readings_columns:
            if col in fragment:
                data_dict[col].append(fragment[col])
            else:
                data_dict[col].append([])

    df = pd.DataFrame(data_dict)
    return df

def read_and_clean_data(file_path):
    with open(file_path, 'r') as file:
        raw_data = file.read()
    clean_data = raw_data.replace(' ', '').replace('\n', '')
    return clean_data

# Example usage
file_path = '../../data/sector_data4_sel.txt'
data = read_and_clean_data(file_path)
parsed_data, readings_columns = parse_data(data)
df = to_dataframe(parsed_data, readings_columns)
df
```

```python
import matplotlib.pyplot as plt

def plot_readings(df, row_number, reading_column):
    readings = df.iloc[row_number][reading_column]
    plt.figure(figsize=(10, 5))
    plt.plot(range(len(readings)), readings, marker='o')
    plt.title(f'Readings from {reading_column} (Row {row_number})')
    plt.xlabel('Index')
    plt.ylabel('Value')
    plt.grid(True)
    plt.show()

plot_readings(df, 2, 3)
```

```python
def plot_readings_grid(df, reading_column, start_row, num_rows):
    fig, axes = plt.subplots(4, 4, figsize=(20, 15))
    axes = axes.flatten()
    
    for i in range(num_rows):
        row_number = start_row + i
        readings = df.iloc[row_number][reading_column]
        axes[i].plot(range(len(readings)), readings, marker='o')
        axes[i].set_title(f'Row {row_number}')
        axes[i].set_xlabel('Index')
        axes[i].set_ylabel('Value')
        axes[i].grid(True)
    
    plt.tight_layout()
    plt.show()

# Plot a 4x4 grid of subplots for readings_3, rows 1 to 16
plot_readings_grid(df, 3, start_row=1, num_rows=16)
```
